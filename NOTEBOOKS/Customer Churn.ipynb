# NO OF CUSTOMER WHO WNATS TO CHURN OR NOT:

```python
sns.countplot(x="Churn", data=df)

plt.title("Customer Churn Distribution")
plt.show()
```

# CHURN BY GENDER:

```python
sns.countplot(x="gender", hue="Churn", data=df)
plt.title("Churn by Gender")
plt.show()
```

# CHURN BY LONG-TERM VS SHORT-TERM:

```python
sns.countplot(x="Contract", hue="Churn", data=df)
plt.xticks(rotation=15)
plt.title("Churn by Contract Type")
plt.show()
```

```python
sns.barplot(x="Churn",y="MonthlyCharges",data=df)
plt.title("Monthly-Charges vs Churn")
plt.show()
```

# TENURE COMPARISON:

```python
sns.boxplot(x="Churn", y="tenure", data=df)

plt.title("Tenure vs Churn")
plt.show()
```

## Business Insights:

- Customers with month-to-month contracts show higher churn rates.
- Long-term contract customers are less likely to churn.
- Customers with lower tenure tend to churn more frequently.
- Higher monthly charges are associated with increased churn.
- Customer retention improves with longer service duration.
